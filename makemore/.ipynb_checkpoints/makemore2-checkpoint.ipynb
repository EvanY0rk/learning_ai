```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt # for making figures
%matplotlib inline
```

```python
#read in all the words
words=open('names.txt', 'r').read().splitlines()
words[:8]
```

```python
len(words)
```

```python
#build the vocabulary of characters and mapping to/from integers
chars=sorted(list(set(''.join(words))))
stoi={s:i+1 for i,s in enumerate(chars)}
stoi['.']=0
itos={i:s for s,i in stoi.items()}
print(itos)
```

```python
#builds the dataset

block_size = 3 #context langth: how many characters so we take to predict the next one?
X, Y = [], []

for w in words:


    context=[0]*block_size
    for ch in w + '.':
        ix=stoi[ch]
        X.append(context)
        Y.append(ix)
        context=context[1:] +[ix] #crop and append


X=torch.tensor(X)
Y=torch.tensor(Y)
```

```python
g=torch.Generator().manual_seed(2147483647) # fpr reproducibility
C=torch.randn((27, 2), generator=g)
W1=torch.randn((6,100), generator=g)
b1=torch.randn(100, generator=g)
W2=torch.randn((100,27), generator=g)
b2=torch.randn(27, generator=g)
parameters=[C, W1, b1, W2, b2]
```

```python
sum(p.nelement() for p in parameters) #number of parameter in total
```

```python
for p in parameters:
    p.requires_grad=True
```

```python
lre=torch.linspace(-3, 0, 1000)
lrs=10**lre

```

```python
lri=[]
lossi=[]
for i in  range(1000000):
    # minibach constructa
    ix=torch.randint(0, X.shape[0], (32,))
    
    #forward pass
    emb=C[X[ix]] #(32, 3, 2)
    h=torch.tanh(emb.view(-1, 6) @ W1+b1) #(32, 100)
    logits=h @ W2+b2 #(32, 27)
    #counts=logits.exp()
    #prob=counts / counts.sum(1, keepdims=True)
    #loss=-prob[torch.arange(32), Y].log().mean()
    loss=F.cross_entropy(logits,Y[ix])


    
    #backward pass
    for p in parameters:
        p.grad=None
    loss.backward()
    
    #update
    #lr = lrs[i]
    lr=0.1
    for p in parameters:
        p.data += -lr *p.grad

    #track stats
    #lri.append(lre[i])
    #lossi.append(loss.item())

print(loss.item())
```

```python
plt.plot(lri,lossi)
```
